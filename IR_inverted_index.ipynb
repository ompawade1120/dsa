```python
import re  # Import regular expression module
```

```python
# Dictionary containing sample documents
documents = {
    "Doc1": "Python is easy to learn",
    "Doc2": "Python is used for Artificial Intelligence",
    "Doc3": "Artificial Intelligence makes life easy",
    "Doc4": "Machine Learning uses Python"
}
```

```python

# Empty dictionary for storing inverted index
inverted_index = {}
```

```python
# Visit each document one by one
for doc_id, text in documents.items():

    # Convert text to lowercase and extract only words
    words = re.findall(r'\w+', text.lower())

    # Read every word in the document
    for word in words:

        # Create a new entry if the word is not present
        if word not in inverted_index:
            inverted_index[word] = []

        # Store document ID only once
        if doc_id not in inverted_index[word]:
            inverted_index[word].append(doc_id)
```

```python
# Print the complete inverted index
print("Inverted Index\n")
for word in sorted(inverted_index):
    print(f"{word:15} -> {inverted_index[word]}")
```

```python
# Accept search query
query = input("Enter word to search: ").lower()

# Retrieve matching documents
if query in inverted_index:
    print("\nDocuments Found:")
    for doc in inverted_index[query]:
        print(doc)
else:
    print("No matching document found.")
```
