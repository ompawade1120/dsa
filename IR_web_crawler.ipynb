```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
#Step 2: Fetch website HTML
url = "http://books.toscrape.com/"
response = requests.get(url)
print("Status Code:", response.status_code)
```

```python
#Step 3: Parse HTML &amp; Find Products
soup = BeautifulSoup(response.text, "html.parser")

#Find all product blocks
products = soup.find_all("article",class_="product_pod")
print("Total products found on page : ", len(products))
```

```python
#Step 4: Extract Product Info
data = []
for product in products:
    title = product.h3.a["title"] #Extract book title
    price = product.find("p",class_="price_color").text #Extract price
    link = url + product.h3.a["href"] #Extract link
    data.append({"Title":title,"Price":price,"Link":link})

df = pd.DataFrame(data)
df.head()
```

```python
#Step 5: Save to CSV
df.to_csv("products.csv", index=False)
print("Product data saved to products.csv")
```
